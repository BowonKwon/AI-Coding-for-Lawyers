## `p01_functions.py` — 4.1 함수: 이름 붙인 작업 묶음

````python
# 4.1 함수: 이름 붙인 작업 묶음

def yearly_interest(principal, rate):        # 매개변수 두 개
    """원금과 연이율로 1년 이자를 계산한다."""   # 설명문(docstring). 없어도 되지만 붙이는 습관
    return principal * rate                   # 결과를 돌려준다

print(yearly_interest(100000000, 0.05))       # 인자 두 개를 넣어 호출
print(yearly_interest(rate=0.03, principal=5000000))   # 이름을 붙여 순서 무관

def interest(principal, rate=0.05, years=1):  # 기본값이 있는 매개변수
    return principal * rate * years

print(interest(100000000))                    # rate, years는 기본값
print(interest(100000000, years=3))           # 일부만 지정

def split_case_no(case_no):                   # 값을 여러 개 돌려주기 (튜플)
    return case_no[:4], case_no[4:6], case_no[6:]

year, code, number = split_case_no("2099가합1234")
print(year, code, number)

def describe(case_no, *tags):                 # 개수가 정해지지 않은 인자
    print(case_no, "태그:", tags)

describe("2099가합1234")
describe("2099가합1234", "손해배상", "항소")
````

````python
# 돌려주는 값이 없는 함수는 None을 돌려준다
def greet(name):
    print(f"{name}님, 안녕하세요")

result = greet("홍길동")
print(result)
````

````python
# 함수 안에서 만든 변수는 밖에서 보이지 않는다 (지역 변수)
def make_total(amounts):
    total = 0
    for a in amounts:
        total += a
    return total

print(make_total([1, 2, 3]))
try:
    print(total)
except NameError as e:
    print("밖에서는 못 씀:", e)
````

# 4.2 모듈: 이 파일 자체가 모듈입니다. 다른 파일에서 import caselib 로 불러 씁니다.

def parse_case_no(case_no):
    """'2099가합1234' -> (2099, '가합', 1234)"""
    year = int(case_no[:4])
    number = int(case_no[6:])
    code = case_no[4:6]
    return year, code, number


def total_claim(records):
    """사건 목록의 청구금액 합계"""
    return sum(r["claim"] for r in records)


def count_by(records, key):
    """records를 key 값별로 센다. 예: count_by(records, 'type')"""
    counts = {}
    for r in records:
        counts[r[key]] = counts.get(r[key], 0) + 1
    return counts

## `p03_files.py` — 4.3 파일 읽고 쓰기 (텍스트, CSV, JSON)

````python
# 4.3 파일 읽고 쓰기 (텍스트, CSV, JSON)
import csv
import json
````

````python
# ---- 텍스트 파일 쓰기와 읽기
with open("memo.txt", "w", encoding="utf-8") as f:     # w: 새로 쓰기 (있으면 덮어씀)
    f.write("첫 줄\n")
    f.write("둘째 줄\n")

with open("memo.txt", "r", encoding="utf-8") as f:     # r: 읽기
    text = f.read()
print(repr(text))                                       # repr: 줄바꿈 문자까지 보이게

with open("memo.txt", encoding="utf-8") as f:           # 한 줄씩
    for line in f:
        print("줄:", line.strip())
````

````python
# ---- CSV: 표 형식. 엑셀에서 열리는 가장 단순한 형식
records = [
    {"case_no": "2099가합1001", "type": "손해배상", "claim": 420000000},
    {"case_no": "2099가합1002", "type": "대여금", "claim": 150000000},
]
with open("cases.csv", "w", encoding="utf-8-sig", newline="") as f:   # utf-8-sig: 엑셀 한글 깨짐 방지
    writer = csv.DictWriter(f, fieldnames=["case_no", "type", "claim"])
    writer.writeheader()
    writer.writerows(records)

with open("cases.csv", encoding="utf-8-sig", newline="") as f:
    loaded = list(csv.DictReader(f))
print(loaded)
print(type(loaded[0]["claim"]))                          # CSV에서 읽은 값은 전부 글자!
print(sum(int(r["claim"]) for r in loaded))              # 숫자로 바꿔 합계
````

````python
# ---- JSON: 딕셔너리·리스트를 그대로 저장하는 형식 (설정 파일, 결과 저장에 씀)
with open("cases.json", "w", encoding="utf-8") as f:
    json.dump(records, f, ensure_ascii=False, indent=2)  # ensure_ascii=False: 한글 그대로

with open("cases.json", encoding="utf-8") as f:
    loaded2 = json.load(f)
print(loaded2[0]["claim"], type(loaded2[0]["claim"]))    # JSON은 숫자를 숫자로 보존
````

## `p04_exceptions.py` — 4.4 예외 처리: 오류가 나도 프로그램이 죽지 않게

````python
# 4.4 예외 처리: 오류가 나도 프로그램이 죽지 않게

amount_texts = ["1200000", "350,000", "미정", "800000"]
````

````python
# 처리하지 않으면 두 번째 값에서 프로그램이 멈춥니다
total = 0
for t in amount_texts:
    try:
        total += int(t)
    except ValueError:                        # 숫자로 못 바꾸는 경우만 잡는다
        print(f"건너뜀: {t!r}")               # !r: repr 형태로 출력
print("합계:", total)
````

````python
# 쉼표 제거를 먼저 시도하고, 그래도 안 되면 건너뛰기
def to_int(text):
    try:
        return int(text.replace(",", ""))
    except ValueError:
        return None

print([to_int(t) for t in amount_texts])
````

````python
# 여러 종류의 예외, else, finally
def read_first_line(path):
    try:
        f = open(path, encoding="utf-8")
    except FileNotFoundError:
        print("파일이 없습니다:", path)
        return None
    else:                                     # 예외가 없었을 때만
        line = f.readline()
        f.close()
        return line.strip()
    finally:                                  # 예외가 있든 없든 항상
        print("read_first_line 끝")

print(read_first_line("memo.txt"))
print(read_first_line("없는파일.txt"))
````

````python
# 직접 예외 일으키기: 잘못된 입력을 초기에 걸러낸다
def parse_case_no(case_no):
    if len(case_no) < 7 or not case_no[:4].isdigit():
        raise ValueError(f"사건번호 형식이 아닙니다: {case_no!r}")
    return int(case_no[:4]), case_no[4:6], int(case_no[6:])

print(parse_case_no("2099가합1234"))
try:
    parse_case_no("가합1234")
except ValueError as e:
    print("오류:", e)
````

## `p05_classes.py` — 4.5 클래스와 객체: 데이터와 동작을 한 묶음으로

````python
# 4.5 클래스와 객체: 데이터와 동작을 한 묶음으로

class Case:
    """사건 하나를 나타내는 틀(클래스)"""

    def __init__(self, case_no, claim, result=None):   # 객체를 만들 때 자동 실행
        self.case_no = case_no                          # self.이름 = 이 객체가 가진 값(속성)
        self.claim = claim
        self.result = result

    def year(self):                                     # 메서드: 이 객체가 할 수 있는 동작
        return int(self.case_no[:4])

    def is_decided(self):
        return self.result is not None

    def __repr__(self):                                 # print 할 때 보이는 모양
        return f"Case({self.case_no}, {self.claim:,}, {self.result})"


c1 = Case("2099가합1001", 420000000, "인용")   # 객체 만들기 = 클래스 이름을 함수처럼 호출
c2 = Case("2099가합1002", 150000000)          # result는 기본값 None
print(c1)
print(c1.case_no, c1.year(), c1.is_decided())
print(c2.is_decided())
c2.result = "기각"                            # 속성은 나중에 바꿀 수 있다
print(c2)
````

````python
# ---- 상속: 기존 틀을 물려받아 조금만 바꾸기
class AppealCase(Case):                       # Case의 모든 것을 물려받는다
    def __init__(self, case_no, claim, result=None, lower_case_no=None):
        super().__init__(case_no, claim, result)       # 부모의 __init__ 먼저
        self.lower_case_no = lower_case_no

    def is_decided(self):                     # 부모 메서드를 덮어쓰기(override)
        return self.result in ("항소기각", "취소", "변경")


a1 = AppealCase("2099나5001", 420000000, "항소기각", lower_case_no="2099가합1001")
print(a1, a1.year(), a1.is_decided(), a1.lower_case_no)
````

````python
# ---- 특수 메서드: 2부의 Dataset이 쓰는 두 가지
class CaseList:
    def __init__(self, cases):
        self.cases = cases

    def __len__(self):                        # len(obj) 가 호출한다
        return len(self.cases)

    def __getitem__(self, i):                 # obj[i] 가 호출한다
        return self.cases[i]


cl = CaseList([c1, c2, a1])
print(len(cl), cl[0], cl[-1])
for c in cl:                                  # __getitem__ 이 있으면 for 문도 된다
    print(" ", c.case_no)
````

## `case_report.py` — 4.6 종합 예제: CSV에서 사건을 읽어 객체로 만들고 보고서를 출력한다

이 파일은 준비 셀에서 이미 받아 두었습니다. 내용은 아래와 같습니다. (왼쪽 폴더 모양 아이콘에서 파일을 두 번 누르면 열어서 고칠 수도 있습니다.)

```python
# 4.6 종합 예제: CSV에서 사건을 읽어 객체로 만들고 보고서를 출력한다
# 3장의 case_records.py 를 함수·모듈·파일·예외·클래스로 다시 쓴 것.
# 실행: python case_report.py cases_all.csv   (파일이 없으면 만들어 줍니다)
import csv
import sys

from caselib import count_by                  # 4.2에서 만든 모듈 재사용


class Case:
    def __init__(self, case_no, case_type, claim, result, days):
        self.case_no = case_no
        self.type = case_type
        self.claim = claim
        self.result = result
        self.days = days

    @classmethod
    def from_row(cls, row):
        """CSV 한 행(딕셔너리, 값은 전부 글자)에서 Case를 만든다. 형식이 틀리면 ValueError."""
        return cls(
            row["case_no"],
            row["type"],
            int(row["claim"].replace(",", "")),
            row["result"],
            int(row["days"]),
        )

    def __repr__(self):
        return f"Case({self.case_no}, {self.type}, {self.claim:,}, {self.result}, {self.days}일)"


SAMPLE = [
    ["case_no", "type", "claim", "result", "days"],
    ["2099가합1001", "손해배상", "420000000", "인용", "310"],
    ["2099가합1002", "대여금", "150,000,000", "기각", "198"],      # 쉼표 있는 금액
    ["2099가합1003", "손해배상", "880000000", "일부인용", "455"],
    ["2099가합1004", "부당이득", "67000000", "인용", "120"],
    ["2099가합1005", "손해배상", "230000000", "기각", "260"],
    ["2099가합1006", "손해배상", "미정", "인용", "90"],            # 잘못된 행: 건너뛰어야 함
]


def ensure_sample(path):
    """예제 파일이 없으면 만든다."""
    try:
        open(path, encoding="utf-8-sig").close()
    except FileNotFoundError:
        with open(path, "w", encoding="utf-8-sig", newline="") as f:
            csv.writer(f).writerows(SAMPLE)
        print(f"예제 파일을 만들었습니다: {path}")


def load_cases(path):
    cases, skipped = [], []
    with open(path, encoding="utf-8-sig", newline="") as f:
        for row in csv.DictReader(f):
            try:
                cases.append(Case.from_row(row))
            except (ValueError, KeyError) as e:
                skipped.append((row.get("case_no", "?"), str(e)))
    return cases, skipped


def report(cases):
    records = [c.__dict__ for c in cases]     # 객체 → 딕셔너리 (caselib 함수에 넘기기 위해)
    print(f"사건 {len(cases)}건, 청구금액 합계 {sum(c.claim for c in cases):,}원")
    print("유형별 건수:", count_by(records, "type"))
    print("결과별 건수:", count_by(records, "result"))
    longest = max(cases, key=lambda c: c.days)
    print(f"최장 처리: {longest.case_no} ({longest.days}일)")


if __name__ == "__main__":
    path = sys.argv[1] if len(sys.argv) > 1 else "cases_all.csv"   # 명령줄 인자
    ensure_sample(path)
    cases, skipped = load_cases(path)
    report(cases)
    for case_no, why in skipped:
        print(f"건너뜀 {case_no}: {why}")
```
