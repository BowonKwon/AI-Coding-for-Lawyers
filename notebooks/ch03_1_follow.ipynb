## `p02_variables.py` — 3.3 변수와 자료형

```python
# 3.3 변수와 자료형

case_no = "2099가합1234"      # 문자열(str): 따옴표로 감싼 글자
claim = 350000000             # 정수(int): 청구금액 3억 5천만 원
rate = 0.05                   # 실수(float): 연 이율 5%
is_final = False              # 불리언(bool): 확정 여부. True 또는 False만 가능

print(case_no, claim, rate, is_final)
print(type(case_no), type(claim), type(rate), type(is_final))
```

```python
# 계산과 갱신
interest = claim * rate       # 1년치 이자
print("1년 이자:", interest)
claim = claim + 10000000      # 변수는 다시 채울 수 있습니다 (청구취지 확장)
print("확장 후 청구금액:", claim)
```

```python
# 형 변환
amount_text = "1200000"       # 서류에서 읽어 온 금액은 글자입니다
print(amount_text + "원")     # 글자끼리는 이어 붙습니다
print(int(amount_text) + 1)   # 숫자로 바꿔야 계산됩니다
print(str(claim) + "원")      # 숫자를 글자로
```

```python
# f-string: 글자 안에 변수 값을 끼워 넣기
print(f"{case_no} 사건의 청구금액은 {claim:,}원, 이자는 {interest:,.0f}원입니다.")
```

## `p04_collections.py` — 3.5~3.8 묶음 자료형: 리스트, 튜플, 딕셔너리, 집합

```python
# 3.5~3.8 묶음 자료형: 리스트, 튜플, 딕셔너리, 집합

# ---- 리스트: 순서가 있고 바꿀 수 있는 묶음
parties = ["원고", "피고"]
parties.append("보조참가인")          # 끝에 추가
print(parties, len(parties))
print(parties[0], parties[-1])        # 첫째, 마지막
print(parties[1:])                    # 둘째부터 끝까지
parties[1] = "피고 주식회사 가나"      # 바꾸기
print(parties)
print("원고" in parties)              # 포함 여부

amounts = [3000, 1200, 5400, 800]
print(sorted(amounts))                # 정렬한 새 리스트 (원본은 그대로)
print(sum(amounts), max(amounts), min(amounts))
```

```python
# ---- 튜플: 순서가 있지만 바꿀 수 없는 묶음
judgment_date = (2026, 9, 6)
year, month, day = judgment_date      # 언패킹: 한 번에 꺼내기
print(f"{year}. {month}. {day}. 선고")
```

```python
# ---- 딕셔너리: 이름표(키)로 값을 찾는 묶음
case = {
    "case_no": "2099가합1234",
    "court": "가상지방법원",
    "claim": 350000000,
    "parties": ["원고", "피고"],      # 값으로 리스트도 들어갑니다
}
print(case["court"])                  # 키로 값 찾기
case["result"] = "일부 인용"          # 새 키 추가
print(case.get("judge", "미입력"))    # 없는 키는 get으로 안전하게
print(list(case.keys()))              # 키 목록
for key, value in case.items():       # 키와 값을 함께 순회
    print(f"  {key}: {value}")
```

```python
# ---- 집합: 중복이 없고 순서도 없는 묶음
cited_a = {"민법 제750조", "민법 제751조", "민사소송법 제202조"}
cited_b = {"민법 제750조", "상법 제401조"}
print(cited_a & cited_b)              # 교집합: 둘 다 인용한 조문
print(cited_a | cited_b)              # 합집합
print(cited_a - cited_b)              # 차집합: A에만 있는 조문
print(len(set(["민법 제750조", "민법 제750조", "민법 제751조"])))  # 중복 제거
```

## `p05_control.py` — 3.9~3.10 조건문과 반복문

```python
# 3.9~3.10 조건문과 반복문
# 아래 기준 금액과 이율은 설명을 위한 가정값입니다. 실제 규정과 다릅니다.

# ---- 조건문
claim = 350000000
if claim > 500000000:
    print("A 유형")
elif claim > 100000000:
    print("B 유형")
else:
    print("C 유형")

is_final = False
has_appeal = True
if not is_final and has_appeal:       # and, or, not 으로 조건을 잇습니다
    print("상소심 계속 중")
```

```python
# ---- for: 묶음을 하나씩 꺼내며 반복
amounts = [3000, 1200, 5400, 800]
total = 0
for a in amounts:
    total = total + a
print("합계:", total)

for i, a in enumerate(amounts):       # 번호와 값을 함께
    print(f"{i + 1}번째: {a}")

for n in range(3):                    # 0, 1, 2
    print("range:", n)
```

```python
# ---- 조건과 반복을 함께
big = []
for a in amounts:
    if a >= 3000:
        big.append(a)
print("3000 이상:", big)
```

```python
# 같은 일을 한 줄로 (리스트 컴프리헨션)
print("3000 이상:", [a for a in amounts if a >= 3000])
```

```python
# ---- while: 조건이 참인 동안 반복
balance = 1000
years = 0
while balance < 2000:
    balance = balance * 1.05          # 가정 이율 5%
    years += 1                        # years = years + 1 과 같습니다
print(f"{years}년 뒤 {balance:.0f}")
```

```python
# ---- break, continue
for a in amounts:
    if a < 1000:
        continue                      # 이번 차례는 건너뜀
    if a > 5000:
        break                         # 반복을 아예 끝냄
    print("처리:", a)
```

## `case_records.py` — 3.11 종합 예제: 사건 목록 집계

```python
# 3.11 종합 예제: 사건 목록 집계
# 아래 데이터는 설명을 위해 지어낸 가상의 사건입니다.

records = [
    {"case_no": "2099가합1001", "type": "손해배상", "claim": 420000000, "result": "인용",    "days": 310},
    {"case_no": "2099가합1002", "type": "대여금",   "claim": 150000000, "result": "기각",    "days": 198},
    {"case_no": "2099가합1003", "type": "손해배상", "claim": 880000000, "result": "일부인용", "days": 455},
    {"case_no": "2099가합1004", "type": "부당이득", "claim": 67000000,  "result": "인용",    "days": 120},
    {"case_no": "2099가합1005", "type": "손해배상", "claim": 230000000, "result": "기각",    "days": 260},
]
```

```python
# 1) 전체 건수와 청구금액 합계
total_claim = 0
for r in records:
    total_claim += r["claim"]
print(f"사건 {len(records)}건, 청구금액 합계 {total_claim:,}원")
```

```python
# 2) 사건 유형별 건수
count_by_type = {}
for r in records:
    t = r["type"]
    count_by_type[t] = count_by_type.get(t, 0) + 1
print("유형별 건수:", count_by_type)
```

```python
# 3) 결과별 평균 처리일수
days_by_result = {}
for r in records:
    days_by_result.setdefault(r["result"], []).append(r["days"])
for result, days in days_by_result.items():
    print(f"{result}: 평균 {sum(days) / len(days):.0f}일 ({len(days)}건)")
```

```python
# 4) 조건에 맞는 사건 고르기: 손해배상이면서 청구금액 3억 원 이상
selected = [r["case_no"] for r in records if r["type"] == "손해배상" and r["claim"] >= 300000000]
print("손해배상·3억 이상:", selected)
```

```python
# 5) 청구금액 순 정렬 (큰 순)
for r in sorted(records, key=lambda r: r["claim"], reverse=True):
    print(f"  {r['case_no']}  {r['claim']:>13,}원  {r['result']}")
```
